```python
import nbformat
from pathlib import Path

# Ruta completa a la carpeta TRAIN con los notebooks de Kaggle
carpeta = Path(r"C:\Users\leo23\OneDrive\Documentos\TIC\git\LLaMA2_7b_emotion_detector\notebooks\TRAIN")

# Buscar todos los archivos .ipynb en esa carpeta
notebooks = sorted(carpeta.glob("*.ipynb"))

print(f"Se encontraron {len(notebooks)} notebooks para limpiar.\n")

for nb_path in notebooks:
    print(f"Limpieza de metadatos en: {nb_path.name}")

    with open(nb_path, "r", encoding="utf-8") as f:
        nb = nbformat.read(f, as_version=4)

    # Eliminar 'widgets' en metadata del notebook si existe
    if "widgets" in nb.metadata:
        print("  → Eliminando 'metadata.widgets'")
        del nb.metadata["widgets"]

    # Eliminar 'widgets' en metadata de cada celda si existe
    for cell in nb.cells:
        if "widgets" in cell.get("metadata", {}):
            print("  → Eliminando 'cell.metadata.widgets'")
            del cell["metadata"]["widgets"]

    # Guardar notebook limpio (sobrescribir)
    with open(nb_path, "w", encoding="utf-8") as f:
        nbformat.write(nb, f)

print("\n✅ Limpieza completada. Metadatos problemáticos eliminados, salidas conservadas.")
```
